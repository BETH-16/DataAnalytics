# Exercise 1.C Performing operations on arrays

```python
import numpy as np
```

## Sample Arrays

```python
r_1 = np.random.randint(1, 20, size=2)
r_2 = np.random.randint(1, 20, size=(1, 2))
r_3 = np.random.random(size=(2, 1))
r_4 = np.random.randint(20, 100, size=(4, 5))
r_5 = np.random.random(size=(4, 5))

print('r_1:', r_1)
print('r_2:', r_2)
print('r_3:', r_3)
print('r_4:', r_4)
print('r_5:', r_5)
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
# shuffle() rearranges the ROWS of the array randomly
# but keeps each row intact — elements within each row stay in the same order
```

The shuffle function rearranged the rows of r_4 randomly but kept the elements within each row in the same order. Only the row positions changed, not the individual values inside each row.

```python
new_shuffle = np.random.permutation(r_4)
print('new_shuffle:')
print(new_shuffle)
print('r_4 after permutation:')
print(r_4)
# permutation() returns a NEW shuffled array without changing the original
# shuffle() changes the original array in place
```

Use shuffle() when you want to rearrange the original array permanently. Use permutation() when you want a shuffled copy but need to keep the original array unchanged.

## Mathematical Operations

```python
print('r_1:', r_1, '| shape:', r_1.shape)
print('r_2:', r_2, '| shape:', r_2.shape)
result = np.add(r_1, r_2)
print('r_1 + r_2:', result)
print('shape:', result.shape)
```

```python
print('r_1:', r_1, '| shape:', r_1.shape)
print('r_3:', r_3, '| shape:', r_3.shape)
result = np.add(r_1, r_3)
print('r_1 + r_3:', result)
print('shape:', result.shape)
```

```python
print('r_2:', r_2, '| shape:', r_2.shape)
print('r_4:', r_4, '| shape:', r_4.shape)
result = np.add(r_2, r_4)
print('r_2 + r_4:', result)
print('shape:', result.shape)
```

```python
try:
    print(np.add(r_1, r_4))
except ValueError as e:
    print(f'r_1 + r_4 Error: {e}')

try:
    print(np.add(r_2, r_4))
except ValueError as e:
    print(f'r_2 + r_4 Error: {e}')

try:
    print(np.add(r_3, r_4))
except ValueError as e:
    print(f'r_3 + r_4 Error: {e}')
# r_1 has shape (2,) and r_4 has shape (4,5) — not compatible
# r_3 has shape (2,1) and r_4 has shape (4,5) — not compatible
```

```python
print(np.add(np.array([1]), r_4))
# adds 1 to every element in r_4
```

```python
print(np.add(np.array([[1]]), r_4))
# adds 1 to every element in r_4 — same result as above
```

```python
print(np.add(np.array([1,1,1,1,1]), r_4))
# adds 1 to each column of every row in r_4
```

```python
try:
    print(np.add(np.array([[1],[1]]), r_4))
except ValueError as e:
    print(f'Error: {e}')
# Error: shape (2,1) cannot broadcast with shape (4,5)
# 2 rows cannot broadcast to 4 rows
```

Broadcasting allows NumPy to add arrays of different shapes by stretching the smaller array to match the larger one. A single value gets added to every element. A row of values gets added to each matching column across all rows. The shape of the result matches the larger array. Broadcasting only works when the dimensions are compatible — either equal or one of them is 1.

## Rounding

```python
print(r_3)
```

```python
print(np.trunc(r_3))
# trunc cuts off the decimal — always toward zero
```

```python
print(np.fix(r_3))
# fix works like trunc — rounds toward zero
```

```python
print(np.floor(r_3))
# floor always rounds DOWN to nearest integer
```

```python
print(np.ceil(r_3))
# ceil always rounds UP to nearest integer
```

```python
print(np.round(r_3, 2))
# round rounds to 2 decimal places
```

## Cumulative Operations

```python
print(r_4)
```

```python
print(np.sum(r_4))
```

```python
print(np.sum(r_4, axis=0))
# sums each column across all rows
```

```python
print(np.sum(r_4, axis=1))
# sums each row across all columns
```

```python
print(np.cumsum(r_4))
# rolling total of all values flattened
```

```python
print(np.cumsum(r_4, axis=0))
# rolling total down each column
```

## Grocery Problem

```python
# bag1: 2 vegetables, 3 fruits, 4 snacks, 0 beverages
# bag2: 4 vegetables, 0 fruits, 1 snack, 2 beverages
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print('bag1:', bag1)
print('bag2:', bag2)
```

```python
total_bag1 = np.sum(bag1)
total_bag2 = np.sum(bag2)
print(f'The first bag contains {total_bag1} items, and the second bag contains {total_bag2} items.')
```

```python
totals = np.add(bag1, bag2)
print(f'There are {totals[0]} vegetables, {totals[1]} fruits, {totals[2]} boxes of snacks, and {totals[3]} beverages in the two grocery bags.')
```

```python
import numpy as np
import pandas as pd
```

## Series

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(list(ac))
ser4 = pd.Series(ad)

print(ser1)
print(ser2)
print(ser3)
print(ser4)
```

```python
try:
    ser5 = pd.Series(17, 18, 19, 20)
except TypeError as e:
    print(f'TypeError: {e}')
# TypeError: Series() only accepts one positional argument
# To create a Series from literal values use a list: pd.Series([17, 18, 19, 20])
```

## DataFrames from Series

```python
df1 = pd.DataFrame(ser1)
df1
```

```python
df1.insert(1, 'ser2', ser2)
df1.insert(2, 'ser3', ser3)
df1
```

```python
df1['ser4'] = ser4
df1
# ser4 was created from a dictionary which has string keys (first, second, third, fourth)
# The other Series used numeric indexes (0,1,2,3)
# Because the indexes don't match, ser4 values show as NaN in the DataFrame
```

```python
df1.loc[4] = [5, 9, 12, 0]
df1
```

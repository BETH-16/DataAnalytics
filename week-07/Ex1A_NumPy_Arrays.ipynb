# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
# Create arrA as a 2D array
arrA = np.array([(11, 12, 13), (14, 15, 16)])
print(arrA)
```

```python
# Loop through the first dimension of arrA
for x in arrA:
    print(x * 2)
```

```python
# Loop through the second dimension of arrA
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop goes through each row in arrA and prints the entire row multiplied by 2, so it outputs two arrays.The second loop goes one level deeper, iterating through each element within the rows and printing each value multiplied by 2, resulting in six individual numbers.

```python
# Loop through second dimension and assign x*2 to loop_result
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)
# loop_result stores x*2 (the entire row multiplied by 2) not y*2
# x is the full row, y is each individual element
```

## Joining and Stacking Arrays

```python
# Concatenate arrA with arrA on axis=0 (default)
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
# Concatenate arrA with arrA on axis=1
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
# Stack joined_array_1 with joined_array_1 on default axis
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# Stack joined_array_1 with joined_array_2 on default axis
try:
    stacked_array_2 = np.stack((joined_array_1, joined_array_2))
    print(stacked_array_2)
except ValueError as e:
    print(f'Error: {e}')
# Error occurs because joined_array_1 and joined_array_2 have different shapes
# np.stack() requires all input arrays to have the same shape
```

```python
# Concatenate joined_array_1 with arrA
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
# Concatenate joined_array_1 with arrA on axis=1
try:
    another_2 = np.concatenate((joined_array_1, arrA), axis=1)
    print(another_2)
except ValueError as e:
    print(f'Error: {e}')
# Error occurs because joined_array_1 has 4 rows and arrA has only 2 rows
# When joining on axis=1 the number of rows must match
```

```python
# hstack - stacks horizontally (adds columns)
hstacked = np.hstack((arrA, arrA))
print('hstack:')
print(hstacked)

# vstack - stacks vertically (adds rows)
vstacked = np.vstack((arrA, arrA))
print('vstack:')
print(vstacked)

# dstack - stacks depth-wise (adds a third dimension)
dstacked = np.dstack((arrA, arrA))
print('dstack:')
print(dstacked)
```

 ## Splitting Arrays

```python
# Split joined_array_1 into 2
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# Result: 2 arrays, each with 2 rows and 3 columns (2D arrays)
```

```python
# Split joined_array_1 into 3
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# Result: 3 arrays — first two have 2 rows, last one has 0 rows
# joined_array_1 has 4 rows so it can't split evenly into 3
```

## Searching and Replacing Array Values

```python
# Use where() to find even numbers in arrA
# If even print the value, if odd print 'odd'
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
# Create arrB as a 2D array
arrB = np.array([(7, 6), (13, 5), (7, 3)])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
# Print arrB first
print(arrB)

# Filter values less than 7
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

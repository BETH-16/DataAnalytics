# Exercise 1.B NumPy Random

```python
import numpy as np
```

 ## Generating Random Numbers

```python
# Random set of integers between 1 and 20 with size of 2 elements
r_1 = np.random.randint(1, 20, size=2)
print(r_1)
```

```python
# Random set of integers between 1 and 20 with size of 1 row, 2 columns
r_2 = np.random.randint(1, 20, size=(1, 2))
print(r_2)
```

```python
# Random set of floats between 0 and 1 with size of 2 rows, 1 column
r_3 = np.random.random(size=(2, 1))
print(r_3)
```

```python
# Random set of integers between 20 and 100 with size of 4 rows, 5 columns
r_4 = np.random.randint(20, 100, size=(4, 5))
print(r_4)
```

```python
# Random set of floats between 0 and 1 with size of 4 rows, 5 columns
r_5 = np.random.random(size=(4, 5))
print(r_5)
```

## Generating Random Choices from an Array

```python
# Create a 1D array of 12 classmate names
students = np.array(['Sayed', 'Sahar', 'luis', 'Priya', 'Jordan', 
                     'samia', 'hana', 'dagi', 'Grace', 'Ken', 
                     'JB', 'charles'])
print(students)
```

```python
# Choose one random selection from students
call_on = np.random.choice(students)
print(call_on)
```

```python
# Choose random sampling from students - 3 rows, 4 columns
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
# This would NOT be an effective way to assign breakout groups because
# np.random.choice() allows repeats by default — the same student
# could appear in multiple groups at the same time!
# A better approach would be to use np.random.shuffle() instead
```

```python
# Choose 5 random selections with different probability values
call_on_weighted = np.random.choice(students, size=5, p=[
    0.30,  # Sayed - 30% chance
    0.10,  # Sahar - 10% chance
    0.10,  # luis - 10% chance
    0.10,  # Priya - 10% chance
    0.10,  # Jordan - 10% chance
    0.05,  # samia - 5% chance
    0.05,  # hana - 5% chance
    0.05,  # dagi - 5% chance
    0.05,  # Grace - 5% chance
    0.05,  # Ken - 5% chance
    0.00,  # JB - 0% chance (will never be called on)
    0.05   # charles - 5% chance
])
print(call_on_weighted)
# p values must add up to exactly 1.0 (100%)
# JB has 0% so will never be selected
# Sayed has 30% so is most likely to be called on
```

## Generating a Normal Distribution

```python
# Random normal distribution
# mean=10, standard deviation=5, size=4 rows x 5 columns
bell_curve = np.random.normal(10, 5, size=(4, 5))
print(bell_curve)
```
